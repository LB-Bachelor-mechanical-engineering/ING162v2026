## Oppgave 1: Simulering av kule, strikk og gass

### 1a) Hvorfor trengs bare én momentumbalanse?

Systemet består av en kule, en strikk og en gass-sylinder. I oppgaveteksten spesifiseres det at både strikken og sylinderen/gassen er masseløse. Siden momentum er definert som $P = m \cdot v$, vil objekter uten masse ikke ha noe momentum å balansere. Det er kun kulen som har masse ($m=20$ kg), og derfor er det kun dens bevegelse vi trenger en balanselov for.

### 1b) Konstitutive relasjoner i skriptet
De konstitutive relasjonene beskriver materialegenskapene og hvordan kreftene oppstår. I Python-skriptet finner vi disse på:
- **Linje 38:** Kraften fra strikken ($F_{band}$), som følger Hookes lov.
- **Linje 39:** Kraften fra gassen ($F_{gas}$), som følger idealgassloven ($p = \frac{NRT}{V}$).

### 1c) Fortegn på arbeid (Linje 42)
Linje 42 i skriptet er: `Phi = Q + dot(-Fgas, v)`.
Her representerer `dot(-Fgas, v)` den mekaniske effekten på gassen. Siden $F_{gas}$ er kraften gassen utøver på kulen, vil $-F_{gas}$ være kraften kulen utøver på gassen (Newtons 3. lov). Når denne ganges med hastigheten $v$, får vi arbeidet utført på gassen per tidsenhet. Dette øker gassens indre energi $E$.

### 1d) Energiinnhold i gass vs. strikk
Gassen modelleres med en indre energi $E$ som avhenger av temperatur ($E = C \cdot N \cdot T$), og denne oppdateres i hver tidssteg. Strikken derimot, har ingen egen energivariabel i dette skriptet; dens energi behandles kun indirekte gjennom kraften den utøver på kulen.

## Oppgave 2: Gummikule som spretter

**Konklusjon:** Output 2 er mest korrekt.
**Begrunnelse:** Mekanisk arbeid er definert ved $W = \int \vec{F} \cdot d\vec{r}$. ## Oppgave 3: Basketball med spinn

For å løse denne oppgaven må vi bruke fundamentale balanselover for mekanikk og termodynamikk:
1. **Newtons andre lov (Impulsbalanse):**
   Kraften fra bakken på ballen er relatert til endringen i ballens bevegelsesmengde (momentum).
   $$\vec{F}_{netto} = \frac{d\vec{p}}{dt}$$
   Her må vi huske at tyngdekraften ($F_g = m \cdot g$) også virker på ballen i negativ $z$-retning.

2. **Mekanisk Effekt ($P_{mek}$):**
   Effekt er arbeid utført per tidsenhet. For en kraft som virker på en flate som glir, er den mekaniske effekten fra kontaktpunktet gitt ved:
   $$P = \vec{F} \cdot \vec{v}_{kontakt}$$
   hvor $\vec{v}_{kontakt}$ er glidehastigheten i kontaktpunktet.

3. **Energibalanse (Termodynamikkens første lov):**
   Total energi i ballen består av kinetisk energi ($K$), potensiell energi ($V$) og indre energi ($U$). Endringen i total energi må tilsvare netto effekt inn i systemet (mekanisk effekt minus avgitt varme):
   $$\frac{dE_{tot}}{dt} = \frac{dK}{dt} + \frac{dV}{dt} + \frac{dU}{dt} = P_{inn} - \dot{Q}_{ut}$$

```python
# --- Gitte verdier ---
m = 0.60          # kg
v_glide = 5.0     # m/s (i x-retning)
dpz_dt = 900.0    # N (tidsderivert av z-momentum)
dK_dt = 9000.0    # J/s (rate kinetisk energi øker)
dU_dt = -12500.0  # J/s (rate indre energi avtar)
dV_dt = 0.0       # J/s (potensiell energi endrer seg ikke)
Q_ut = 330.0      # J/s (avgitt varme)
g = 9.8           # N/kg

# --- (a) Finn kraften fra bakken (Fx og Fz) ---

# I z-retning: dpz/dt = Fz_bakke - m*g
# Fz_bakke = dpz/dt + m*g
Fz_bakke = dpz_dt + (m * g)

# For å finne Fx må vi se på energibalansen.
# Energibalanse: dK/dt + dV/dt + dU/dt = P_mek - Q_ut
# P_mek er her effekten fra kontaktkraften i x-retning: Fx * v_glide
# (Fz gjør ikke arbeid siden det ikke er vertikal bevegelse i kontaktpunktet)

# Vi løser for P_mek først:
P_mek = dK_dt + dV_dt + dU_dt + Q_ut

# Siden P_mek = Fx_bakke * v_glide (kraft i samme retning som glidehastighet):
# Merk: Bakken bremser ballen, så Fx vil virke mot glidehastigheten.
Fx_bakke = P_mek / v_glide

# --- Resultater ---
print("--- Oppgave 3 Resultater ---")
print(f"(a) Kraft fra bakken i z-retning (Fz): {Fz_bakke:.2f} N")
print(f"(a) Kraft fra bakken i x-retning (Fx): {Fx_bakke:.2f} N")
print(f"(b) Mekanisk effekt inn i ballen (P_mek): {P_mek:.2f} J/s")

print("\n--- Forklaring (c) ---")
print("1. Bevegelsesmengdebalanse (Newtons 2. lov) brukes for å finne Fz.")
print("2. Energibalansen (1. hovedsetning) brukes for å koble endring i energi")
print("   (kinetisk, indre, potensiell) med arbeid (mekanisk effekt) og varme.")
print("3. Konstitutiv relasjon for effekt (P = F * v) brukes for å koble kraft og hastighet.")
```

## Oppgave 4: Biljardkollisjon

For å verifisere om kollisjonsdataene er fysisk mulige, må vi sjekke tre fundamentale konserveringslover:
1. **Lineær bevegelsesmengde ($\vec{p} = m\vec{v}$):** Siden det ikke er noen netto ytre kraft, må $\sum \vec{p}_{før} = \sum \vec{p}_{etter}$.
2. **Dreieimpuls ($\vec{L} = I_c\vec{\omega}$):** Siden netto ytre dreiemoment er null, må $\sum \vec{L}_{før} = \sum \vec{L}_{etter}$.
3. **Total energi ($E = K_{trans} + K_{rot} + U$):** Ifølge arbeid-energi-teoremet må den totale mekaniske energien (inkludert indre energi $U$) være konstant når det ikke gjøres ytre arbeid.

```python
import numpy as np

# --- Konstanter fra oppgaven ---
m = 0.1600  # kg
Ic = 5.198e-5  # kg m^2

# --- Data: Før kollisjon (t0) ---
v1_for = np.array([0.000, 1.000, 0.000])
w1_for = np.array([-35.09, 0.000, -350.9])
u1_for = 0.000

v2_for = np.array([0.000, 0.000, 0.000])
w2_for = np.array([0.000, 0.000, 0.000])
u2_for = 0.000

# --- Data: Etter kollisjon (t1) ---
v1_etter = np.array([-1.046, -0.1706, 0.000])
w1_etter = np.array([5.987, -36.70, -163.1])
u1_etter = 1.025

v2_etter = np.array([1.046, 1.171, 0.000])
w2_etter = np.array([-41.07, 36.70, 80.47])
u2_etter = 1.025

def beregn_statistikk(v1, w1, u1, v2, w2, u2):
    # Bevegelsesmengde p = m * v
    p_tot = m * v1 + m * v2
    
    # Dreieimpuls L = Ic * w
    l_tot = Ic * w1 + Ic * w2
    
    # Kinetisk energi (translasjon + rotasjon) + indre energi
    k_trans = 0.5 * m * np.sum(v1**2) + 0.5 * m * np.sum(v2**2)
    k_rot = 0.5 * Ic * np.sum(w1**2) + 0.5 * Ic * np.sum(w2**2)
    e_tot = k_trans + k_rot + u1 + u2
    
    return p_tot, l_tot, e_tot

# Utfører beregninger
p_for, l_for, e_for = beregn_statistikk(v1_for, w1_for, u1_for, v2_for, w2_for, u2_for)
p_etter, l_etter, e_etter = beregn_statistikk(v1_etter, w1_etter, u1_etter, v2_etter, w2_etter, u2_etter)

# --- Presentasjon av resultater ---
print(f"{'Størrelse':<25} | {'Før kollisjon':<30} | {'Etter kollisjon':<30}")
print("-" * 90)

print(f"{'Bevegelsesmengde (p)':<25} | {str(p_for):<30} | {str(p_etter):<30}")
print(f"{'Dreieimpuls (L)':<25} | {str(l_for):<30} | {str(l_etter):<30}")
print(f"{'Total energi (E)':<25} | {e_for:<30.4f} J | {e_etter:<30.4f} J")

print("\n--- Kontroll (Sjekker 3 signifikante sifre) ---")
def sjekk_bevaring(navn, for_verdi, etter_verdi):
    diff = np.abs(for_verdi - etter_verdi)
    # Enkel sjekk på relativ differanse eller absolutt for små tall
    tol = 1e-3 
    if np.all(diff < tol):
        print(f"[OK] {navn} er bevart.")
    else:
        print(f"[FEIL] {navn} er IKKE bevart!")

sjekk_bevaring("Bevegelsesmengde", p_for, p_etter)
sjekk_bevaring("Dreieimpuls", l_for, l_etter)
sjekk_bevaring("Energi", e_for, e_etter)
```
